## Problem 1: Exploring Petal Length

### Using the Iris dataset, investigate the distribution of petal_length.

The first thing needed would be to import the packages we need.

```python
#Importing the sklearn Package - which contains the Iris Dataset
import sklearn as skl

#This is the quickest way to get a normal distribution calculation. Only added this later while researching
from scipy.stats import norm

#Imported Pandas
import pandas as pd

#Imported Numpy to use.
import numpy as np

# Imported matplotlib
import matplotlib.pyplot as plt
```

```python
# Next, we're going to need our Iris Dataset
iris = skl.datasets.load_iris()
iris
```

```python
# Next, we transform this into a proper dataframe
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df.head()
```

```python
# Then, we take the Petal Length
petal_length = df['petal length (cm)']
```

Now that we've extracted what we needed, we're going to plot our histogram.

```python
# We need to define the number of bins. I went with the square root method mainly because we had more bars. the Friedman Draconis made too few bars.
bin_edges = np.histogram_bin_edges(petal_length, bins='sqrt')

# Then we plot our histogram for Petal_Lengths
plt.hist(petal_length, bins=bin_edges, edgecolor='black')

plt.xlabel('Petal Length (cm)')
plt.ylabel('Frequency')
plt.title('Distribution of Iris Petal Lengths')

plt.show()

# References:
# Discussions on number of bins - https://stats.stackexchange.com/questions/798/calculating-optimal-number-of-bins-in-a-histogram
# Numpy Documentation - https://numpy.org/doc/stable/reference/generated/numpy.histogram_bin_edges.html
```

Next, we're going to superimpose the normal or standard distribution

```python
# Numpy already provides us with these functions
mean = np.mean(petal_length)
standard_deviation = np.std(petal_length)

print(f"Mean: {mean:.2f}")
print(f"Standard Deviation: {standard_deviation:.2f}")
```

Now that we have those values calculated, let's try and superimpose a normal distribution. We already imported our automatic norm calculator from scipy

```python

# We first create values as needed as per the scipy documentation.
x = np.linspace(petal_length.min(), petal_length.max(), 100) 

# This calculates the normal distribution with the values we already from before.
y = norm.pdf(x, mean, standard_deviation)

# We plot the histogram
plt.hist(petal_length, bins=bin_edges, density=True, edgecolor='black')

# And then superimpose the normal curve
plt.plot(x, y, color='red', linewidth=2, label='Normal Distribution')

plt.xlabel('Petal Length (cm)')
plt.ylabel('Probability Density')
plt.title('Distribution of Iris Petal Lengths')
plt.legend()

plt.show()

# References
# https://docs.scipy.org/doc/scipy-1.18.0/reference/generated/scipy.stats.norm.html#scipy.stats.norm - Calculates the normal distribution automatically.
# https://stackoverflow.com/a/10138381 - Retrieved 2026-11-05, License - CC BY-SA 4.0 - Where I found the Norm function
```

The above doesn't seem normal. It's as if the curve does not fit what's happening with the data, but the value are correct, but the mean and
standard deviation values are correct.

Next, let's plot histograms for each of the three species and superimpose the normal distribution on each of them.

```python
# Let's first get the species. This is already included in the dataset.

df['species'] = iris.target_names[iris.target]
df.head()
```

```python
# Next, we separate them and get the petal length for each species

setosa = df[df['species'] == 'setosa']['petal length (cm)']
versicolor = df[df['species'] == 'versicolor']['petal length (cm)']
virginica = df[df['species'] == 'virginica']['petal length (cm)']

# Need to also calculate new bins since we actually have less values instead of the full 150
setosa_bins = np.histogram_bin_edges(setosa, bins='sqrt')
versicolor_bins = np.histogram_bin_edges(versicolor, bins='sqrt')
virginica_bins = np.histogram_bin_edges(virginica, bins='sqrt')
```

Let's reuse our code from before, for each of the three species. Let's calculate the mean and standard distribution for each species and make our plots.

### Setosa

```python

# Calculation of the values
mean_setosa = np.mean(setosa)
standard_deviation_setosa = np.std(setosa)

print(f"Mean: {mean_setosa:.2f}")
print(f"Standard Deviation: {standard_deviation_setosa:.2f}")

# Create the line
x_setosa = np.linspace(setosa.min(), setosa.max(), 100)

# Calculate the normal distribution for Setosa
y_setosa = norm.pdf(x_setosa, mean_setosa, standard_deviation_setosa)

# Superimposed Plot
plt.hist(setosa, bins='sqrt', density=True, edgecolor='black')
plt.plot(x_setosa, y_setosa, color='red', linewidth=2,label='Normal Distribution')

plt.xlabel('Petal Length (cm)')
plt.ylabel('Probability Density')
plt.title('Distribution of Setosa Petal Lengths')
plt.legend()

plt.show()
```

### Versicolor

```python
# Do the same for Versicolor

mean_versicolor = np.mean(versicolor)
standard_deviation_versicolor = np.std(versicolor)

print(f"Mean: {mean_versicolor:.2f}")
print(f"Standard Deviation: {standard_deviation_versicolor:.2f}")


x_versicolor = np.linspace(versicolor.min(), versicolor.max(), 100)
y_versicolor = norm.pdf(x_versicolor, mean_versicolor, standard_deviation_versicolor)


plt.hist(versicolor, bins='sqrt', density=True, edgecolor='black')
plt.plot(x_versicolor, y_versicolor, color='red', linewidth=2,label='Normal Distribution')

plt.xlabel('Petal Length (cm)')
plt.ylabel('Probability Density')
plt.title('Distribution of Versicolor Petal Lengths')
plt.legend()

plt.show()
```

### Virginica

```python
# And the same for Virginica

mean_virginica = np.mean(virginica)
standard_deviation_virginica = np.std(virginica)

print(f"Mean: {mean_virginica:.2f}")
print(f"Standard Deviation: {standard_deviation_virginica:.2f}")


x_virginica = np.linspace(virginica.min(), virginica.max(), 100)
y_virginica = norm.pdf(x_virginica, mean_virginica, standard_deviation_virginica)


plt.hist(virginica, bins='sqrt', density=True, edgecolor='black')
plt.plot(x_virginica, y_virginica, color='red', linewidth=2,label='Normal Distribution')

plt.xlabel('Petal Length (cm)')
plt.ylabel('Probability Density')
plt.title('Distribution of Versicolor Petal Lengths')
plt.legend()

plt.show()
```
